```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("creditcard.csv")
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes.value_counts())

print("\nNon-numeric columns:")
print(df.select_dtypes(exclude=np.number).columns.tolist())

print("\nDuplicate column names:", df.columns.duplicated().sum())
```

```python
print("Missing values:", df.isna().sum().sum())
print("\nMissing values per column:")
display(df.isna().sum().sort_values(ascending=False).head(10))

numeric_df = df.select_dtypes(include=np.number)

print("Infinite values:", np.isinf(numeric_df).sum().sum())
```

```python
print(df["Class"].value_counts())
print("\nClass proportions:")
print(df["Class"].value_counts(normalize=True).mul(100).round(4))

fraud_count = (df["Class"] == 1).sum()
legitimate_count = (df["Class"] == 0).sum()

print("Fraud cases:", fraud_count)
print("Legitimate cases:", legitimate_count)
print("Imbalance ratio:", legitimate_count / fraud_count)

df["Class"].value_counts().sort_index().plot(
    kind="bar", rot=0
)
plt.title("Class Distribution")
plt.xlabel("Class (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Number of Transactions")
plt.show()
```
